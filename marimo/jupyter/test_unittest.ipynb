```python
from pathlib import Path

import pandas as pd
```

```python
import unittest


def divide(a: int, b: int) -> float:
    return a / b

class stefan_zhelev_test(unittest.TestCase):
    def test_division(self):
        self.assertEqual(divide(10, 2), 5)

    def test_division_again(self):
        self.assertTrue(divide(10, 2) == 5.0)

    def test_data_type_of_division_result(self):
        self.assertTrue(type(divide(10, 2)) is float)

unittest.main(argv=[""], exit=False)
```
